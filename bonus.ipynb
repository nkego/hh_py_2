```python
import pandas as pd
from sqlalchemy import create_engine

# параметры подключения к бд
DB_USER = 'postgres'
DB_PASSWORD = '123'
DB_HOST = 'localhost'
DB_PORT = '5432'
DB_NAME = 'postgres'



# Формируем строку подключения (URI)
db_url = f"postgresql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}"

# Создаем движок для подключения
engine = create_engine(db_url)


schema = 'hh_school'
tables_to_read = ['areas', 'employers', 'vacancies', 'vacancy_details', 'responses']

data = {}

for table_name in tables_to_read:
    query = f"SELECT * FROM {schema}.{table_name}"
        
    data[table_name] = pd.read_sql(query, con=engine)
    print(f" Таблица '{table_name}' загружена. Размер: {data[table_name].shape}")


display(data['vacancies'].head())
```
